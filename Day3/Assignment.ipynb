```python
import tiktoken
from transformers import AutoTokenizer
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# 1. CONFIGURATION
# ============================================================

# Hugging Face tokenizer
QWEN_MODEL = "Qwen/Qwen3-0.6B"

# Sentence Transformer model
EMBEDDING_MODEL = "all-MiniLM-L6-v2"


# ============================================================
# 2. PREDEFINED SENTENCES
# ============================================================

sentences = [
    "I enjoy building software and writing code.",
    "Machine learning and artificial intelligence are changing technology.",
    "The weather is very cold today.",
    "I love playing football with my friends.",
    "Python is a popular programming language."
]


# ============================================================
# 3. GET USER INPUT
# ============================================================

text = input("\nEnter your text: ").strip()

if not text:
    print("Please enter some text.")
    exit()


# ============================================================
# 4. TIKTOKEN TOKENIZATION
# ============================================================

print("\n" + "=" * 70)
print("TIKTOKEN TOKENIZATION")
print("=" * 70)

# GPT-style tokenizer
encoding = tiktoken.get_encoding("cl100k_base")

tiktoken_ids = encoding.encode(text)
tiktoken_tokens = [encoding.decode([token_id]) for token_id in tiktoken_ids]

print(f"\nToken Count: {len(tiktoken_ids)}")

print("\nTokens:")
for i, token in enumerate(tiktoken_tokens):
    print(f"{i + 1:2}. {repr(token)}")

print("\nToken IDs:")
print(tiktoken_ids)


# ============================================================
# 5. HUGGING FACE / QWEN TOKENIZATION
# ============================================================

print("\n" + "=" * 70)
print("HUGGING FACE / QWEN TOKENIZATION")
print("=" * 70)

qwen_tokenizer = AutoTokenizer.from_pretrained(QWEN_MODEL)

qwen_ids = qwen_tokenizer.encode(
    text,
    add_special_tokens=False
)

qwen_tokens = qwen_tokenizer.convert_ids_to_tokens(qwen_ids)

print(f"\nToken Count: {len(qwen_ids)}")

print("\nTokens:")
for i, token in enumerate(qwen_tokens):
    print(f"{i + 1:2}. {repr(token)}")

print("\nToken IDs:")
print(qwen_ids)


# ============================================================
# 6. SIDE-BY-SIDE TOKEN COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("TOKENIZER COMPARISON")
print("=" * 70)

print(f"\n{'TIKTOKEN':<35} {'QWEN':<35}")
print("-" * 70)

max_length = max(len(tiktoken_tokens), len(qwen_tokens))

for i in range(max_length):

    if i < len(tiktoken_tokens):
        t_token = repr(tiktoken_tokens[i])
        t_id = tiktoken_ids[i]
        t_value = f"{t_token} ({t_id})"
    else:
        t_value = ""

    if i < len(qwen_tokens):
        q_token = repr(qwen_tokens[i])
        q_id = qwen_ids[i]
        q_value = f"{q_token} ({q_id})"
    else:
        q_value = ""

    print(f"{t_value:<35} {q_value:<35}")

print("-" * 70)

print(f"Tiktoken token count : {len(tiktoken_tokens)}")
print(f"Qwen token count     : {len(qwen_tokens)}")


# ============================================================
# 7. LOAD SENTENCE TRANSFORMER
# ============================================================

print("\n" + "=" * 70)
print("SEMANTIC SIMILARITY")
print("=" * 70)

print("\nLoading SentenceTransformer model...")

embedding_model = SentenceTransformer(EMBEDDING_MODEL)


# ============================================================
# 8. CREATE EMBEDDING FOR USER INPUT
# ============================================================

user_embedding = embedding_model.encode(
    [text],
    normalize_embeddings=True
)


# ============================================================
# 9. CREATE EMBEDDINGS FOR PREDEFINED SENTENCES
# ============================================================

sentence_embeddings = embedding_model.encode(
    sentences,
    normalize_embeddings=True
)


# ============================================================
# 10. CALCULATE COSINE SIMILARITY
# ============================================================

similarity_scores = cosine_similarity(
    user_embedding,
    sentence_embeddings
)[0]


# ============================================================
# 11. CREATE RANKING
# ============================================================

results = list(zip(sentences, similarity_scores))

results.sort(
    key=lambda x: x[1],
    reverse=True
)


# ============================================================
# 12. DISPLAY RESULTS
# ============================================================

print("\nUser Input:")
print(f"\"{text}\"")

print("\nSemantic Similarity Ranking:")
print("-" * 70)

for rank, (sentence, score) in enumerate(results, start=1):

    print(
        f"{rank}. Score: {score:.4f} | {sentence}"
    )

print("-" * 70)

print("\nMost semantically similar sentence:")
print(f"\"{results[0][0]}\"")
print(f"Similarity Score: {results[0][1]:.4f}")
```
