```python
import tiktoken
from transformers import AutoTokenizer

# model for HuggingFace
model = 'Qwen/Qwen2-0.5B'

# tiktoken
encoding_tool = 'cl100k_base'

# Encode the text
text = "I love Programming"

def hugginface_tokenizer(text):
  tokenizer = AutoTokenizer.from_pretrained(model, trust_remote_code=True)
  token_ids = tokenizer.encode(text)
  print(f'AutoTransformer Tokens: ', token_ids)

  # Decode the tokens
  decoded_text = tokenizer.decode(token_ids)
  print(f'Decoded Text: {decoded_text}')

  # Length of tokens
  token_count = len(token_ids)
  print(f'Token Count: {token_count}')

def tiktoken_tokenizer(text):
  encoder = tiktoken.get_encoding(encoding_tool)

  # Encode the text
  tokens = encoder.encode(text)

  # Decode the text
  decoded_text = encoder.decode(tokens)
  print(f'Decoded Text: {decoded_text}')

  print(f'Tokens: {tokens}')
  token_count = len(tokens)
  print(f'Token Count: {token_count}')

tiktoken_tokenizer(text)
hugginface_tokenizer(text)
```
