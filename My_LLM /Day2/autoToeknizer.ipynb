````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")
text = "I love hyderabad biryani!!"

tokens = tokenizer.tokenize(text)
tokens_id = tokenizer.encode(text)

print(f"Original text:", {text})
print(f"Tokens: {tokens}")
print(f"Tokens ID: {tokens_id}")
print(f"Decoded text: {tokenizer.decode(tokens_id)}")

````

Comparing Hugging Face `AutoTokenizer` with `tiktoken`.

## Tokenization — Quick Notes

### 1. Hugging Face `AutoTokenizer`

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

text = "I love hyderabad biryani!!"

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("Text:", text)
print("Tokens:", tokens)
print("IDs:", token_ids)
print("Decoded:", tokenizer.decode(token_ids))
```

### What happens?

```text
Text
 ↓
Tokenizer
 ↓
Tokens → ["I", "Ġlove", "Ġhy", ...]
 ↓
Token IDs → [40, 1842, ...]
 ↓
Model
```

* `AutoTokenizer` → loads the tokenizer associated with a Hugging Face model.
* `tokenize()` → gives **token pieces as strings**.
* `encode()` → converts text into **integer token IDs**.
* `decode()` → converts token IDs back to text.
* `Ġ` in GPT-2 tokens represents a **preceding space**.
* GPT-2 uses a **BPE-based tokenizer**.

---

## 2. `tiktoken`

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")

text = "I love hyderabad biryani!!"

token_ids = encoding.encode(text)

print("IDs:", token_ids)
print("Decoded:", encoding.decode(token_ids))
```

* `tiktoken` → tokenizer library commonly used for OpenAI model encodings.
* `encode()` → gives **token IDs**.
* `decode()` → converts IDs back to text.
* To see individual token pieces:

```python
for token_id in token_ids:
    print(token_id, repr(encoding.decode([token_id])))
```

---

## 3. Main comparison

|               | Hugging Face        | tiktoken               |
| ------------- | ------------------- | ---------------------- |
| Library       | `transformers`      | `tiktoken`             |
| Load          | `from_pretrained()` | `get_encoding()`       |
| Token strings | `tokenize()`        | Decode individual IDs  |
| Token IDs     | `encode()`          | `encode()`             |
| Text from IDs | `decode()`          | `decode()`             |
| Common use    | Hugging Face models | OpenAI model encodings |
| Tokenizer     | Model-specific      | Encoding-specific      |

### ⚠️ Most important point

**GPT-2 tokenizer ≠ `cl100k_base` tokenizer.**

The same text can produce **different tokens and different IDs**.

Always use the tokenizer/encoding appropriate for the model.

### LLM flow to remember

```text
Text
 ↓
Tokenizer
 ↓
Token IDs
 ↓
Embeddings
 ↓
Transformer
 ↓
Output
```

**Tokenizer converts human-readable text into numbers that the LLM can process.**
